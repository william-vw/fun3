```python
import functools

class Memoize():
    
    def __init__(self, func):
        functools.update_wrapper(self, func)
        self.func = func
        self.args_res = {}
        
    def __call__(self, *args):
        if args not in self.args_res:
            self.args_res[args] = self.func(*args)
        
        return self.args_res[args]
```

```python
@Memoize
def test(a, b):
    print("exec", a, b)
    return a * b

test(1, 2)
test(3, 4)
test(1, 2)

test.args_res
```

```python
import functools
import inspect

class MemoizeCtuPass():
    
    def __init__(self, func):
        functools.update_wrapper(self, func)
        self.func = func
        self.args_res = {}
    
    def memoize(self, args, res):
        self.args_res[args] = res
    
    def __call__(self, *args):
        pass_args = args[:-1]
        ctu_fn = args[-1]
        
        if args not in self.args_res:
            # lmbda *cb_args: self.args_res[args] = cb_args
            
            # cannot use assignment expressions when indexing dictionary
            self.func(*pass_args, lambda *cb_args: self.memoize(pass_args, cb_args))
            
        res = self.args_res[pass_args]
        ctu_fn(*res)
```

```python
@MemoizeCtuPass
def test(a, b, final_ctu):
    print("exec")
    final_ctu(a*b, "abc")
    
test(1, 2, lambda x, y: print(x, y))
test(3, 4, lambda x, y: print(x, y))
test(1, 2, lambda x, y: print(x, y))

test.args_res
```
