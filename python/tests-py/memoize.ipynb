```python
import sys
sys.path.insert(0, "..")
from pathlib import Path
from ast import dump, unparse, parse
from n3.ns import xsdNs
from n3.to_py import run_py, save_py
```

```python
# descendantOf example

rules_str = """@prefix : <http://example.org/> . 
@prefix : <http://example.org/> . 

{ ?x :descendantOf ?y } <= { ?x :hasParent ?y } .
{ ?x :descendantOf ?z } <= { ?x :hasParent ?y . ?y :descendantOf ?z } .
"""

query_str = """@prefix : <http://example.org/> .
@prefix : <http://example.org/> . 

?x :descendantOf ?y .
"""

data_str = """@prefix : <http://example.org/> . 
@prefix : <http://example.org/> . 

:will :hasParent :paul .
:paul :hasParent :edward .
# :edward :hasParent :peter . 
"""

params = { 'gen': { 'memoize': True } }
print(run_py(query_str, rules_str, data_str, params=params))
```

```python
# fibonacci example

rules_path = "../tests-misc/fib.n3" 
query_path = "../tests-misc/fib_query.n3"
data_path = "../tests-misc/fib_data.n3"
out_path = "../tests-misc/fib.py"

params = { 'code_dir': "..", 'gen': { 'call_query': True, 'memoize': True } }
save_py(Path(query_path), Path(rules_path), Path(data_path), Path(out_path), params=params)
print(run_py(Path(query_path), Path(rules_path), Path(data_path)))
```
