```python
import os, sys
sys.path.insert(0, "../..")
from n3.parse import parse_n3_file
```

```python
# get the number of triples of the datasets

folder = os.path.join("data")
files = os.listdir(folder)
files.sort()
for file in files:
    if not file.endswith(".n3"):
        continue
    data = parse_n3_file(os.path.join(folder, file)).data
    print(file, len(data))
```
